```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[3,1,5,2,6]

plt.plot(x,y)
plt.title("Python Growth Per month",fontsize=15)
plt.xlabel("Month",fontsize=15)
plt.ylabel("Python",fontsize=15)

plt.text(2,5,"Php",fontsize=15,color="r",)

plt.show()
```
