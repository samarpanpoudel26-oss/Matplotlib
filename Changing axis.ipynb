```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[3,2,1,5,4]

plt.plot(x,y)
plt.xticks(x,labels=["Python","c","c++","php","java"]) #changing axis in x axis
plt.yticks(y) #changing axis in y-axis


plt.show()
```

```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[3,2,1,5,4]

plt.plot(x,y)
plt.xlim(0,10) #changing limit in x axis
plt.ylim(0,10) # changing limit in y-axis

plt.show()
```
