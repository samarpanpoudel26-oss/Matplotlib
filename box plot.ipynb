```python
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70,120]

#patch_artist function is used to fill color
#boxprops is used to change the property of box.It must be passed in the form of dictionary.
#sym is used to change the color of oval.
#capprops is used to change the color of the line.It is also should be passed in the form of dictionary.

plt.boxplot(x,widths=0.13,labels=["Python"],patch_artist=True,showmeans=True,sym='blue',boxprops=dict(color="r"),
            capprops=dict(color="red"),whiskerprops=dict(color="brown"))


plt.show()
```

```python
#inserting multiple box plot
import matplotlib.pyplot as plt

x=[10,20,30,40,50,60,70,120]
x1=[30,10,40,50,60,80,40]
y=[x,x1]



plt.boxplot(y,labels=["Python","Java"],widths=0.13,patch_artist=True,showmeans=True,sym='blue',boxprops=dict(color="r"),
            capprops=dict(color="red"),whiskerprops=dict(color="brown"))


plt.show()
```
