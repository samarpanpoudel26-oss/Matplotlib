```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
y=[1,2,3,4,5]

#mfc=marker face color
plt.step(x,y,marker="o",color="r",mfc="g",label="python")
plt.grid()
plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")


plt.legend(loc=8)
plt.show()
```
