```python
import matplotlib.pyplot as plt

x=[1,2,3,4,5]
area1=[2,5,7,8,3]
area2=[9,1,8,6,3]
area3=[7,5,9,1,5]
l=["area1","area2","area3"]

plt.stackplot(x,area1,area2,area3,labels=l,colors=["r","g","m"])
plt.title("Python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")


            
plt.legend(loc=2)
        
plt.show()
```
