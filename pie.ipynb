```python
import matplotlib.pyplot as plt
x=[10,20,30,40]
y=["c","c++","java","python"]
plt.pie(x,labels=y)
plt.show()
```

```python
#to seperate single part,we use explode parameter
import matplotlib.pyplot as plt

ex=[0.3,0.0,0.0,0.0] #i want to seperate c and 0.4 is the distance
x=[10,20,30,40]
y=["c","c++","java","python"]
c=["b","y","g","r"]

#To show percentage we use autopct function.the formula is"%(number of digit you want to show after decimal)f%%)
#To change the fontsize ,we use textprops function.Value must be passed in dictionary
plt.pie(x,labels=y,explode=ex,colors=c,autopct="%0.1f%%",shadow=True,radius=1.3,labeldistance=1.2,startangle=360,
        textprops={"fontsize":15},wedgeprops={"linewidth":5},center=(0,-3))
plt.title("Popularity")
plt.legend(loc=3)
plt.show()
```
