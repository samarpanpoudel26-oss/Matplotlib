```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(111) #Takind data through numpy array
array=np.random.randint(1,100,50)

plt.title("Popularity",fontsize=15)
plt.xlabel("Python",fontsize=15)
plt.ylabel("No.",fontsize=15)

plt.hist(array,color="Brown",alpha=0.7,edgecolor="red")
plt.show()
```

```python
#Bins parameter is used to seperate the data frames.Now it is quite messey.It gather the data into 1 container deserved like bar graph.
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(111) #Takind data through numpy array
array=np.random.randint(1,100,50)

l=[10,20,30,40,50]
plt.title("Popularity",fontsize=15)
plt.xlabel("Python",fontsize=15)
plt.ylabel("No.",fontsize=15)

plt.hist(array,color="Brown",alpha=0.7,edgecolor="red",bins=l,width=2)
plt.show()
```

```python
#Cumulative=-1 arranges reverse frequency order
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(111) #Takind data through numpy array
array=np.random.randint(1,100,50)

l=[10,20,30,40,50,60]
plt.title("Popularity",fontsize=15)
plt.xlabel("Python",fontsize=15)
plt.ylabel("No.",fontsize=15)

plt.hist(array,color="Brown",alpha=0.7,edgecolor="red",bins=l,cumulative=-1,histtype="step")
plt.show()
```

```python
#Cumulative=-1 arranges reverse frequency order
#To draw a line we can use axvline syntxas
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(111) #Takind data through numpy array
array=np.random.randint(1,100,50)

l=[10,20,30,40,50,60]
plt.title("Popularity",fontsize=15)
plt.xlabel("Python",fontsize=15)
plt.ylabel("No.",fontsize=15)

plt.hist(array,color="Brown",alpha=0.7,edgecolor="red",bins=l,rwidth=0.8)
plt.show()
```
