```python
import matplotlib.pyplot as plt

x=["Python","Java","C++","C#"]
y=[49,41,35,29]
z=[21,23,27,22]

plt.xlabel("Languages",fontsize=15) #labelling x axis
plt.ylabel("Popularity",fontsize=15) #labelling y-axis
plt.title("Languages Popularity",fontsize=20) #labelling title
c=["r",'m','g','y'] #changing the color of the graphs differently

plt.bar(x,y,width=0.5,color=c,align="center",edgecolor="black",linewidth=3,alpha=0.4) #now we are modifying the bar graph
plt.bar(x,z,width=0.5,color="brown",alpha=0.4)#to create double bar graph.But it is overlaped so we need to seperate it.
plt.legend("Popularity")
plt.legend("Popularity1")
plt.show()
```

```python
#seperating the overlapped bar graph
import matplotlib.pyplot as plt
import numpy as np

x=["Python","Java","C++","C#"]
y=[49,41,35,29]
z=[21,23,27,22]

width=0.3
p=np.arange(len(x))
p1=[a+width for a in p]

plt.xlabel("Languages",fontsize=15) #labelling x axis
plt.ylabel("Popularity",fontsize=15) #labelling y-axis
plt.title("Languages Popularity",fontsize=20) #labelling title

c=["r",'m','g','y'] #changing the color of the graphs differently
plt.bar(p1,y,width,color=c,align="center",edgecolor="black",linewidth=3,alpha=0.4) #now we are modifying the bar graph
plt.bar(p,z,width,color="brown",alpha=0.4)#to create double bar graph.But it is overlaped so we need to seperate it.

plt.xticks(p+width/2,x,rotation=20) #restoring the value of x elements
plt.legend("Popularity")
plt.legend("Popularity1")
plt.show()
```

```python
#To create horizontal bar graph ,bar must be barh
import matplotlib.pyplot as plt

x=["Python","Java","C++","C#"]
y=[49,41,35,29]
z=[21,23,27,22]

plt.xlabel("Languages",fontsize=15) #labelling x axis
plt.ylabel("Popularity",fontsize=15) #labelling y-axis
plt.title("Languages Popularity",fontsize=20) #labelling title
c=["r",'m','g','y'] #changing the color of the graphs differently

plt.barh(x,y,color=c,align="center",edgecolor="black",linewidth=3,alpha=0.4) #now we are modifying the bar graph
plt.barh(x,z,color="brown",alpha=0.4)#to create double bar graph.But it is overlaped so we need to seperate it.

plt.legend("Popularity")
plt.legend("Popularity1")
plt.show()
```
